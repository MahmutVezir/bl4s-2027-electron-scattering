```python
from pathlib import Path

PROJECT = Path("/content/electron_scattering")

files = {
"include/PrimaryGeneratorAction.hh": r'''
#ifndef PrimaryGeneratorAction_h
#define PrimaryGeneratorAction_h 1

#include "G4VUserPrimaryGeneratorAction.hh"

class G4ParticleGun;
class G4Event;

namespace B1 {
class PrimaryGeneratorAction : public G4VUserPrimaryGeneratorAction {
public:
    PrimaryGeneratorAction();
    ~PrimaryGeneratorAction() override;

    void GeneratePrimaries(G4Event*) override;

private:
    G4ParticleGun* fParticleGun = nullptr;
};
}

#endif
''',

"src/PrimaryGeneratorAction.cc": r'''
#include "PrimaryGeneratorAction.hh"

#include "G4Event.hh"
#include "G4ParticleGun.hh"
#include "G4ParticleTable.hh"
#include "G4SystemOfUnits.hh"
#include "G4ThreeVector.hh"

namespace B1 {

PrimaryGeneratorAction::PrimaryGeneratorAction()
{
    fParticleGun = new G4ParticleGun(1);

    auto particle =
        G4ParticleTable::GetParticleTable()->FindParticle("e-");

    fParticleGun->SetParticleDefinition(particle);
    fParticleGun->SetParticleEnergy(5.0 * MeV);
    fParticleGun->SetParticlePosition(
        G4ThreeVector(0.0, 0.0, -50.0 * mm)
    );
    fParticleGun->SetParticleMomentumDirection(
        G4ThreeVector(0.0, 0.0, 1.0)
    );
}

PrimaryGeneratorAction::~PrimaryGeneratorAction()
{
    delete fParticleGun;
}

void PrimaryGeneratorAction::GeneratePrimaries(G4Event* event)
{
    fParticleGun->GeneratePrimaryVertex(event);
}

}
''',

"src/DetectorConstruction.cc": r'''
#include "DetectorConstruction.hh"

#include "G4Box.hh"
#include "G4LogicalVolume.hh"
#include "G4NistManager.hh"
#include "G4PVPlacement.hh"
#include "G4SystemOfUnits.hh"
#include "G4ThreeVector.hh"

namespace B1 {

G4VPhysicalVolume* DetectorConstruction::Construct()
{
    auto nist = G4NistManager::Instance();

    auto air = nist->FindOrBuildMaterial("G4_AIR");
    auto aluminium = nist->FindOrBuildMaterial("G4_Al");

    const G4bool checkOverlaps = true;

    auto worldSolid = new G4Box(
        "World", 500.0 * mm, 500.0 * mm, 500.0 * mm
    );

    auto worldLogical = new G4LogicalVolume(
        worldSolid, air, "World"
    );

    auto worldPhysical = new G4PVPlacement(
        nullptr,
        G4ThreeVector(),
        worldLogical,
        "World",
        nullptr,
        false,
        0,
        checkOverlaps
    );

    // 0.1 mm thick aluminium target at z = 0.
    auto targetSolid = new G4Box(
        "Target", 10.0 * mm, 10.0 * mm, 0.05 * mm
    );

    auto targetLogical = new G4LogicalVolume(
        targetSolid, aluminium, "Target"
    );

    new G4PVPlacement(
        nullptr,
        G4ThreeVector(0.0, 0.0, 0.0),
        targetLogical,
        "Target",
        worldLogical,
        false,
        0,
        checkOverlaps
    );

    // Thin air scoring slab at z = 100 mm.
    // Its material is the same as the surrounding world.
    auto planeSolid = new G4Box(
        "ScoringPlane", 100.0 * mm, 100.0 * mm, 0.005 * mm
    );

    auto planeLogical = new G4LogicalVolume(
        planeSolid, air, "ScoringPlane"
    );

    new G4PVPlacement(
        nullptr,
        G4ThreeVector(0.0, 0.0, 100.0 * mm),
        planeLogical,
        "ScoringPlane",
        worldLogical,
        false,
        0,
        checkOverlaps
    );

    return worldPhysical;
}

}
''',

"include/RunAction.hh": r'''
#ifndef RunAction_h
#define RunAction_h 1

#include "G4UserRunAction.hh"
#include "globals.hh"

#include <fstream>

class G4Run;

namespace B1 {

class RunAction : public G4UserRunAction {
public:
    RunAction() = default;
    ~RunAction() override = default;

    void BeginOfRunAction(const G4Run*) override;
    void EndOfRunAction(const G4Run*) override;

    void WriteScatteringRecord(
        G4int eventID,
        G4int trackID,
        G4double energy,
        G4double x,
        G4double y,
        G4double z,
        G4double ux,
        G4double uy,
        G4double uz
    );

    // Retained for compatibility with the unused B1 EventAction.
    void AddEdep(G4double) {}

private:
    std::ofstream fOutput;
};

}

#endif
''',

"src/RunAction.cc": r'''
#include "RunAction.hh"

#include "G4Exception.hh"
#include "G4Run.hh"
#include "G4SystemOfUnits.hh"

#include <iomanip>

namespace B1 {

void RunAction::BeginOfRunAction(const G4Run*)
{
    fOutput.open("electron_scattering.csv", std::ios::out);

    if (!fOutput.is_open()) {
        G4Exception(
            "RunAction::BeginOfRunAction",
            "OutputFileError",
            FatalException,
            "Could not open electron_scattering.csv"
        );
        return;
    }

    fOutput
        << "event_id,track_id,kinetic_energy_MeV,"
        << "x_mm,y_mm,z_mm,ux,uy,uz,theta_deg,phi_deg\n";
}

void RunAction::WriteScatteringRecord(
    G4int eventID,
    G4int trackID,
    G4double energy,
    G4double x,
    G4double y,
    G4double z,
    G4double ux,
    G4double uy,
    G4double uz
)
{
    if (!fOutput.is_open()) return;

    const G4double theta =
        std::atan2(std::sqrt(ux * ux + uy * uy), uz)
        * 180.0 / 3.14159265358979323846;

    const G4double phi =
        std::atan2(uy, ux)
        * 180.0 / 3.14159265358979323846;

    fOutput << std::setprecision(12)
        << eventID << ","
        << trackID << ","
        << energy / MeV << ","
        << x / mm << ","
        << y / mm << ","
        << z / mm << ","
        << ux << ","
        << uy << ","
        << uz << ","
        << theta << ","
        << phi << "\n";
}

void RunAction::EndOfRunAction(const G4Run*)
{
    if (fOutput.is_open()) {
        fOutput.close();
    }
}

}
''',

"include/SteppingAction.hh": r'''
#ifndef SteppingAction_h
#define SteppingAction_h 1

#include "G4UserSteppingAction.hh"

class G4Step;

namespace B1 {

class RunAction;

class SteppingAction : public G4UserSteppingAction {
public:
    explicit SteppingAction(RunAction* runAction);

    void UserSteppingAction(const G4Step*) override;

private:
    RunAction* fRunAction = nullptr;
};

}

#endif
''',

"src/SteppingAction.cc": r'''
#include "SteppingAction.hh"

#include "RunAction.hh"

#include "G4Event.hh"
#include "G4RunManager.hh"
#include "G4Step.hh"
#include "G4StepPoint.hh"
#include "G4StepStatus.hh"
#include "G4TouchableHandle.hh"
#include "G4Track.hh"
#include "G4VPhysicalVolume.hh"

namespace B1 {

SteppingAction::SteppingAction(RunAction* runAction)
    : fRunAction(runAction)
{}

void SteppingAction::UserSteppingAction(const G4Step* step)
{
    auto post = step->GetPostStepPoint();

    if (post->GetStepStatus() != fGeomBoundary) return;

    auto volume = post->GetTouchableHandle()->GetVolume();

    if (!volume || volume->GetName() != "ScoringPlane") return;

    auto track = step->GetTrack();

    // Record only the primary electron, not secondary electrons.
    if (track->GetParentID() != 0) return;
    if (track->GetDefinition()->GetPDGEncoding() != 11) return;

    auto event = G4RunManager::GetRunManager()->GetCurrentEvent();
    if (!event) return;

    auto direction = post->GetMomentumDirection();
    auto position = post->GetPosition();

    fRunAction->WriteScatteringRecord(
        event->GetEventID(),
        track->GetTrackID(),
        post->GetKineticEnergy(),
        position.x(),
        position.y(),
        position.z(),
        direction.x(),
        direction.y(),
        direction.z()
    );
}

}
''',

"src/ActionInitialization.cc": r'''
#include "ActionInitialization.hh"

#include "PrimaryGeneratorAction.hh"
#include "RunAction.hh"
#include "SteppingAction.hh"

namespace B1 {

void ActionInitialization::BuildForMaster() const
{
    SetUserAction(new RunAction);
}

void ActionInitialization::Build() const
{
    SetUserAction(new PrimaryGeneratorAction);

    auto runAction = new RunAction;
    SetUserAction(runAction);

    SetUserAction(new SteppingAction(runAction));
}

}
''',

"exampleB1.cc": r'''
#include "ActionInitialization.hh"
#include "DetectorConstruction.hh"
#include "QBBC.hh"

#include "G4RunManagerFactory.hh"
#include "G4UImanager.hh"

#include <iostream>

int main(int argc, char** argv)
{
    if (argc != 2) {
        std::cerr << "Usage: exampleB1 macro.mac\n";
        return 1;
    }

    auto runManager =
        G4RunManagerFactory::CreateRunManager(G4RunManagerType::Serial);

    runManager->SetUserInitialization(new B1::DetectorConstruction());

    auto physicsList = new QBBC;
    physicsList->SetVerboseLevel(0);
    runManager->SetUserInitialization(physicsList);

    runManager->SetUserInitialization(new B1::ActionInitialization());

    auto uiManager = G4UImanager::GetUIpointer();

    G4String command = "/control/execute ";
    G4String macro = argv[1];

    uiManager->ApplyCommand(command + macro);

    delete runManager;
    return 0;
}
''',

"run_electron.mac": r'''
/run/initialize
/run/printProgress 100
/run/beamOn 1000
'''
}

for relative, content in files.items():
    path = PROJECT / relative
    path.parent.mkdir(parents=True, exist_ok=True)
    path.write_text(content.strip() + "\n")

print("Yazılan dosya sayısı:", len(files))
print("Proje:", PROJECT)
print("Orijinal B1 dosyalarına dokunulmadı.")
```

```python
import numpy as np
import pandas as pd
from pathlib import Path

CSV = Path("/content/electron_scattering/electron_scattering.csv")
df = pd.read_csv(CSV)

required = [
    "event_id", "track_id", "kinetic_energy_MeV",
    "x_mm", "y_mm", "z_mm",
    "ux", "uy", "uz",
    "theta_deg", "phi_deg"
]

assert all(column in df.columns for column in required), "Eksik CSV sütunu."
assert len(df) > 0, "Hiç elektron kaydı yok."
assert df[required].notna().all().all(), "Eksik veya boş veri var."

direction_norm = np.sqrt(
    df["ux"]**2 + df["uy"]**2 + df["uz"]**2
)

print("Kayıt sayısı:", len(df))
print("Benzersiz event sayısı:", df["event_id"].nunique())
print("Kinetik enerji aralığı (MeV):",
      df["kinetic_energy_MeV"].min(),
      df["kinetic_energy_MeV"].max())
print("Ortalama saçılma açısı (derece):", df["theta_deg"].mean())
print("Maksimum saçılma açısı (derece):", df["theta_deg"].max())
print("Yön vektörü normu — ortalama:", direction_norm.mean())
print("Yön vektörü normu — maksimum sapma:",
      np.max(np.abs(direction_norm - 1.0)))

assert (df["kinetic_energy_MeV"] >= 0).all()
assert (df["theta_deg"] >= 0).all()
assert (df["theta_deg"] <= 180).all()
assert np.allclose(direction_norm, 1.0, atol=1e-5)

print("\nTemel veri kontrolleri başarılı.")
```

```python
import numpy as np
import pandas as pd
from pathlib import Path

CSV = Path("/content/electron_scattering/electron_scattering.csv")
df = pd.read_csv(CSV)

required = [
    "event_id", "track_id", "kinetic_energy_MeV",
    "x_mm", "y_mm", "z_mm",
    "ux", "uy", "uz", "theta_deg", "phi_deg"
]

assert all(c in df.columns for c in required)
assert len(df) > 0
assert df[required].notna().all().all()

norm = np.sqrt(df["ux"]**2 + df["uy"]**2 + df["uz"]**2)

print("Kayıt sayısı:", len(df))
print("Benzersiz olay:", df["event_id"].nunique())
print("Ortalama enerji (MeV):", df["kinetic_energy_MeV"].mean())
print("Ortalama saçılma açısı (derece):", df["theta_deg"].mean())
print("Maksimum saçılma açısı (derece):", df["theta_deg"].max())
print("Yön vektörü normu ortalaması:", norm.mean())
print("Yön vektörü normu maksimum sapması:", np.max(np.abs(norm - 1)))

assert (df["kinetic_energy_MeV"] >= 0).all()
assert df["theta_deg"].between(0, 180).all()
assert np.allclose(norm, 1.0, atol=1e-5)

print("\nTemel veri kontrolleri başarılı.")
```

```python
from pathlib import Path

PROJECT = Path("/content/electron_scattering")

detector_code = r'''
#include "DetectorConstruction.hh"

#include "G4Box.hh"
#include "G4LogicalVolume.hh"
#include "G4NistManager.hh"
#include "G4PVPlacement.hh"
#include "G4SystemOfUnits.hh"
#include "G4ThreeVector.hh"

#include <cstdlib>
#include <string>

namespace B1 {

G4VPhysicalVolume* DetectorConstruction::Construct() {
    auto nist = G4NistManager::Instance();

    auto air = nist->FindOrBuildMaterial("G4_AIR");
    auto aluminium = nist->FindOrBuildMaterial("G4_Al");

    const char* setting = std::getenv("SCATTERING_TARGET_MATERIAL");
    const std::string materialName = setting ? setting : "aluminium";

    G4Material* targetMaterial = nullptr;

    if (materialName == "air") {
        targetMaterial = air;
    } else if (materialName == "aluminium") {
        targetMaterial = aluminium;
    } else {
        G4Exception(
            "DetectorConstruction::Construct",
            "InvalidTargetMaterial",
            FatalException,
            "SCATTERING_TARGET_MATERIAL must be air or aluminium."
        );
        return nullptr;
    }

    const G4bool checkOverlaps = true;

    auto worldSolid = new G4Box(
        "World", 500.0 * mm, 500.0 * mm, 500.0 * mm
    );

    auto worldLogical = new G4LogicalVolume(
        worldSolid, air, "World"
    );

    auto worldPhysical = new G4PVPlacement(
        nullptr,
        G4ThreeVector(),
        worldLogical,
        "World",
        nullptr,
        false,
        0,
        checkOverlaps
    );

    auto targetSolid = new G4Box(
        "Target", 10.0 * mm, 10.0 * mm, 0.05 * mm
    );

    auto targetLogical = new G4LogicalVolume(
        targetSolid, targetMaterial, "Target"
    );

    new G4PVPlacement(
        nullptr,
        G4ThreeVector(),
        targetLogical,
        "Target",
        worldLogical,
        false,
        0,
        checkOverlaps
    );

    auto planeSolid = new G4Box(
        "ScoringPlane", 100.0 * mm, 100.0 * mm, 0.005 * mm
    );

    auto planeLogical = new G4LogicalVolume(
        planeSolid, air, "ScoringPlane"
    );

    new G4PVPlacement(
        nullptr,
        G4ThreeVector(0.0, 0.0, 100.0 * mm),
        planeLogical,
        "ScoringPlane",
        worldLogical,
        false,
        0,
        checkOverlaps
    );

    return worldPhysical;
}

}
'''

run_action_code = r'''
#include "RunAction.hh"

#include "G4Exception.hh"
#include "G4Run.hh"
#include "G4SystemOfUnits.hh"

#include <cstdlib>
#include <iomanip>
#include <cmath>

namespace B1 {

void RunAction::BeginOfRunAction(const G4Run*) {
    const char* outputPath = std::getenv("SCATTERING_OUTPUT");
    const char* path = outputPath ? outputPath : "electron_scattering.csv";

    fOutput.open(path, std::ios::out);

    if (!fOutput.is_open()) {
        G4Exception(
            "RunAction::BeginOfRunAction",
            "OutputFileError",
            FatalException,
            "Could not open scattering CSV output."
        );
        return;
    }

    fOutput
        << "event_id,track_id,kinetic_energy_MeV,"
        << "x_mm,y_mm,z_mm,ux,uy,uz,theta_deg,phi_deg\n";
}

void RunAction::WriteScatteringRecord(
    G4int eventID,
    G4int trackID,
    G4double energy,
    G4double x,
    G4double y,
    G4double z,
    G4double ux,
    G4double uy,
    G4double uz
) {
    if (!fOutput.is_open()) return;

    const G4double theta =
        std::atan2(std::sqrt(ux * ux + uy * uy), uz)
        * 180.0 / 3.14159265358979323846;

    const G4double phi =
        std::atan2(uy, ux)
        * 180.0 / 3.14159265358979323846;

    fOutput << std::setprecision(12)
        << eventID << ","
        << trackID << ","
        << energy / MeV << ","
        << x / mm << ","
        << y / mm << ","
        << z / mm << ","
        << ux << ","
        << uy << ","
        << uz << ","
        << theta << ","
        << phi << "\n";
}

void RunAction::EndOfRunAction(const G4Run*) {
    if (fOutput.is_open()) {
        fOutput.close();
    }
}

}
'''

(PROJECT / "src/DetectorConstruction.cc").write_text(detector_code)
(PROJECT / "src/RunAction.cc").write_text(run_action_code)

print("DetectorConstruction.cc güncellendi.")
print("RunAction.cc güncellendi.")
print("Diğer kaynak dosyalarına dokunulmadı.")
```

```python
import subprocess
import os
import json
import hashlib
from pathlib import Path
from datetime import datetime, timezone

PROJECT = Path("/content/electron_scattering")
EXECUTABLE = PROJECT / "build" / "exampleB1"
MACRO = PROJECT / "build" / "run_electron.mac"

BASE = PROJECT / "comparison"
BASE.mkdir(exist_ok=True)

conditions = [
    {
        "name": "control_no_target",
        "material": "air",
        "seed": [12345, 67890]
    },
    {
        "name": "aluminium_target",
        "material": "aluminium",
        "seed": [54321, 9876]
    }
]

results = []

for condition in conditions:
    name = condition["name"]
    output_dir = BASE / name
    output_dir.mkdir(exist_ok=True)

    csv_path = output_dir / "electron_scattering.csv"
    log_path = output_dir / "simulation.log"

    env = os.environ.copy()
    env["SCATTERING_TARGET_MATERIAL"] = condition["material"]
    env["SCATTERING_OUTPUT"] = str(csv_path)

    # Geant4 veri kümelerini Micromamba ortamından al.
    command = [
        "/content/bin/micromamba",
        "run",
        "-p", "/content/geant4-env",
        str(EXECUTABLE),
        str(MACRO)
    ]

    with log_path.open("w") as log:
        process = subprocess.run(
            command,
            cwd=PROJECT,
            env=env,
            stdout=log,
            stderr=subprocess.STDOUT,
            text=True
        )

    record = {
        "condition": name,
        "material": condition["material"],
        "seeds": condition["seed"],
        "return_code": process.returncode,
        "csv_exists": csv_path.exists(),
        "csv_sha256": (
            hashlib.sha256(csv_path.read_bytes()).hexdigest()
            if csv_path.exists() else None
        )
    }

    results.append(record)

    print("\n", name)
    print("Çıkış kodu:", process.returncode)
    print("CSV mevcut:", csv_path.exists())

    if process.returncode != 0:
        print(log_path.read_text(errors="replace")[-4000:])
        raise RuntimeError(f"{name} simülasyonu başarısız.")

    if not csv_path.exists():
        raise FileNotFoundError(f"{name} CSV dosyası oluşmadı.")

metadata = {
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "geant4_version": "11.4.3",
    "primary_particle": "e-",
    "initial_energy_MeV": 5.0,
    "target_thickness_mm": 0.1,
    "scoring_plane_z_mm": 100.0,
    "events_per_condition": 1000,
    "conditions": results
}

metadata_path = BASE / "experiment_metadata.json"
metadata_path.write_text(json.dumps(metadata, indent=2))

print("\nİki koşul da başarıyla tamamlandı.")
print("Metadata:", metadata_path)
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

BASE = Path("/content/electron_scattering/comparison")

control = pd.read_csv(
    BASE / "control_no_target" / "electron_scattering.csv"
)

aluminium = pd.read_csv(
    BASE / "aluminium_target" / "electron_scattering.csv"
)

def summarize(name, df):
    return {
        "condition": name,
        "records": len(df),
        "unique_events": df["event_id"].nunique(),
        "mean_energy_MeV": df["kinetic_energy_MeV"].mean(),
        "std_energy_MeV": df["kinetic_energy_MeV"].std(),
        "mean_theta_deg": df["theta_deg"].mean(),
        "std_theta_deg": df["theta_deg"].std(),
        "median_theta_deg": df["theta_deg"].median(),
        "theta_95_percentile_deg": df["theta_deg"].quantile(0.95),
        "mean_radius_mm": np.sqrt(
            df["x_mm"]**2 + df["y_mm"]**2
        ).mean()
    }

summary = pd.DataFrame([
    summarize("Control: air", control),
    summarize("Aluminium target", aluminium)
])

display(summary)

summary.to_csv(BASE / "comparison_summary.csv", index=False)

print("Karşılaştırma tablosu kaydedildi.")
```
